```python
import pandas as pd

# Load dataset
df = pd.read_csv("battery_vehicle_dataset_300.csv")

# Preview
df.head()
```

```python

# Data Exploration shows data types, missing values, and basic statistics. 
# Check structure
df.info()

# Summary stats
df.describe()

# Unique car makes
df['make'].unique()

# Count of battery types
df['battery_type'].value_counts()
```

```python
# Check missing values
df.isnull().sum()

# Remove duplicates
df = df.drop_duplicates()

# Standardize text
df['make'] = df['make'].str.title()
```

```python
def smart_recommend(make, model):
    result = df[
        (df['make'] == make) &
        (df['model'] == model)
    ]
    
    if result.empty:
        return "No data available"
    
    return result.groupby('battery_type').agg({
        'capacity_ah': 'mean',
        'cca': 'mean'
    })
```

```python
def recommend_battery(make, model, year, engine_cc):
    result = df[
        (df['make'] == make) &
        (df['model'] == model) &
        (df['year'] == year) &
        (df['engine_cc'] == engine_cc)
    ]
    
    if result.empty:
        return "No match found"
    
    return result[[
        'battery_type',
        'battery_group',
        'capacity_ah',
        'voltage',
        'cca'
    ]]
```
